# MotionCare AI — Stream Window Equivalence

Verify streaming batches reproduce offline windows.

Synthetic educational data only. This is not a validated fall detector.

```python
from pathlib import Path
import sys,json
import numpy as np
import pandas as pd
ROOT=Path.cwd()
if not (ROOT/'src').exists():ROOT=ROOT.parent
sys.path.insert(0,str(ROOT/'src'))
from healthcare_ai.session_io import load_session
from healthcare_ai.signal_contracts import CHANNELS
paths=sorted((ROOT/'data').glob('imu_session_*.csv'))
session,fs=load_session(paths[0]);x=session[CHANNELS].to_numpy()
```

```python
from healthcare_ai.stream_buffer import WindowBuffer
from healthcare_ai.windowing import windows
b=WindowBuffer(fs);emitted=[]
for chunk in np.array_split(x,13):emitted.extend(b.push(chunk))
expected=windows(x,fs)
print('Emitted:',len(emitted),'Offline:',len(expected))
print('Equal:',all(np.array_equal(a,w) for a,(_,_,w) in zip(emitted,expected)))
```

## Interpretation

Signals and endpoints are simulated. Patient grouping prevents overlapping windows from crossing partitions. Window-level scores are dependent observations and cannot establish clinical performance.
