# MotionCare AI — Imu Visualization

Visualize simulated acceleration and injected events.

Synthetic educational data only. This is not a validated fall detector.

```python
from pathlib import Path
import sys,json
import numpy as np
import pandas as pd
ROOT=Path.cwd()
if not (ROOT/'src').exists():ROOT=ROOT.parent
sys.path.insert(0,str(ROOT/'src'))
from healthcare_ai.session_io import load_session
from healthcare_ai.signal_contracts import CHANNELS
paths=sorted((ROOT/'data').glob('imu_session_*.csv'))
session,fs=load_session(paths[0]);x=session[CHANNELS].to_numpy()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(figsize=(10,3))
ax.plot(session.seconds,session.ax,label='Acceleration x, arbitrary units')
for event in session.loc[session.event_marker==1,'seconds']:ax.axvline(event,color='crimson',alpha=.6)
ax.set(xlabel='Time (seconds)',ylabel='Arbitrary units',title='Synthetic IMU recording and injected pulse timestamps')
ax.legend();fig.tight_layout()
print('Illustrative simulator signal, not a real patient recording.')
```

## Interpretation

Signals and endpoints are simulated. Patient grouping prevents overlapping windows from crossing partitions. Window-level scores are dependent observations and cannot establish clinical performance.
